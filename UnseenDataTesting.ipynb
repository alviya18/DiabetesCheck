```python
# =========================================================
# DIABETESCHECK
# UNSEEN DATA TESTING - FOUR MODELS
# =========================================================

import pandas as pd
import numpy as np
import joblib
import re

from IPython.display import display

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)


# =========================================================
# 1. LOAD MODELS
# =========================================================

models = {
    "Logistic Regression": joblib.load("logistic_regression.pkl"),
    "Decision Tree": joblib.load("decision_tree.pkl"),
    "Random Forest": joblib.load("random_forest.pkl"),
    "XGBoost": joblib.load("xgboost.pkl")
}

# Scaler used for Logistic Regression
scaler = joblib.load("scaler.pkl")


print("=" * 90)
print("FOUR MODELS LOADED SUCCESSFULLY")
print("=" * 90)


# =========================================================
# 2. LOAD CSV
# =========================================================

csv_file = (
    "DiabetesCheck Unseen Data Collection "
    "(Responses) - Form responses 1.csv"
)

df = pd.read_csv(csv_file)


print("\nCSV loaded successfully.")
print("Number of respondents:", len(df))
print("Number of columns:", len(df.columns))


# =========================================================
# 3. CLEAN COLUMN NAMES
# =========================================================

df.columns = [
    str(col).strip()
    for col in df.columns
]


# =========================================================
# 4. COLUMN FINDER
# =========================================================

def find_column(search_text):

    matches = [
        col
        for col in df.columns
        if search_text.lower() in str(col).lower()
    ]

    if len(matches) == 0:
        raise ValueError(
            f"\nColumn containing '{search_text}' was not found."
        )

    if len(matches) > 1:

        print(
            f"\nMultiple columns matched '{search_text}':"
        )

        for col in matches:
            print(" -", col)

        # Use the first matching column
        return matches[0]

    return matches[0]


# =========================================================
# 5. IDENTIFY CSV COLUMNS
# =========================================================

age_col = find_column(
    "Basic Information"
)

sex_col = find_column(
    "Sex"
)

education_col = find_column(
    "Education level"
)

income_col = find_column(
    "Monthly household income"
)

height_col = find_column(
    "Heght in cm"
)

weight_col = find_column(
    "Weight in kg"
)

highbp_col = find_column(
    "high blood pressure"
)

highchol_col = find_column(
    "high cholesterol"
)

cholcheck_col = find_column(
    "cholesterol checked"
)

smoker_col = find_column(
    "smoked at least 100"
)

stroke_col = find_column(
    "stroke"
)

heart_col = find_column(
    "coronary heart disease"
)

physical_activity_col = find_column(
    "physical activity"
)

fruit_col = find_column(
    "eat fruit"
)

vegetable_col = find_column(
    "eat vegetables"
)

alcohol_col = find_column(
    "heavy amounts of alcohol"
)

healthcare_col = find_column(
    "healthcare coverage"
)

doctor_cost_col = find_column(
    "could not because of cost"
)

general_health_col = find_column(
    "general, how would you rate"
)

mental_health_col = find_column(
    "mental health not good"
)

physical_health_col = find_column(
    "physical health not good"
)

diffwalk_col = find_column(
    "difficulty walking"
)

actual_diabetes_col = find_column(
    "have you ever been told by a healthcare professional "
    "that you have diabetes"
)


# =========================================================
# 6. SHOW IDENTIFIED COLUMNS
# =========================================================

print("\n" + "=" * 90)
print("CSV COLUMNS IDENTIFIED")
print("=" * 90)

column_check = pd.DataFrame({
    "Model Field": [
        "Age",
        "Sex",
        "Education",
        "Income",
        "Height",
        "Weight",
        "HighBP",
        "HighChol",
        "CholCheck",
        "Smoker",
        "Stroke",
        "HeartDiseaseorAttack",
        "PhysActivity",
        "Fruits",
        "Veggies",
        "HvyAlcoholConsump",
        "AnyHealthcare",
        "NoDocbcCost",
        "GenHlth",
        "MentHlth",
        "PhysHlth",
        "DiffWalk",
        "Actual Diabetes"
    ],

    "CSV Column": [
        age_col,
        sex_col,
        education_col,
        income_col,
        height_col,
        weight_col,
        highbp_col,
        highchol_col,
        cholcheck_col,
        smoker_col,
        stroke_col,
        heart_col,
        physical_activity_col,
        fruit_col,
        vegetable_col,
        alcohol_col,
        healthcare_col,
        doctor_cost_col,
        general_health_col,
        mental_health_col,
        physical_health_col,
        diffwalk_col,
        actual_diabetes_col
    ]
})

display(column_check)


# =========================================================
# 7. AGE CONVERSION
# =========================================================

age_mapping = {

    "18–24": 1,
    "18-24": 1,

    "25–29": 2,
    "25-29": 2,

    "30–34": 3,
    "30-34": 3,

    "35–39": 4,
    "35-39": 4,

    "40–44": 5,
    "40-44": 5,

    "45–49": 6,
    "45-49": 6,

    "50–54": 7,
    "50-54": 7,

    "55–59": 8,
    "55-59": 8,

    "60–64": 9,
    "60-64": 9,

    "65–69": 10,
    "65-69": 10,

    "70–74": 11,
    "70-74": 11,

    "75–79": 12,
    "75-79": 12,

    "80+": 13,

    "80 years or older": 13
}


def convert_age(value):

    value = str(value).strip()

    if value in age_mapping:
        return age_mapping[value]

    raise ValueError(
        f"Unknown age category: {value}"
    )


# =========================================================
# 8. SEX CONVERSION
# =========================================================

def convert_sex(value):

    value = str(value).strip().lower()

    if value == "female":
        return 0

    if value == "male":
        return 1

    raise ValueError(
        f"Unknown sex value: {value}"
    )


# =========================================================
# 9. EDUCATION CONVERSION
# =========================================================

education_mapping = {

    "Never attended school / kindergarten": 1,

    "Grades 1–8": 2,
    "Grades 1-8": 2,

    "Grades 9–11": 3,
    "Grades 9-11": 3,

    "High school graduate": 4,

    "Some college / technical school": 5,

    "College graduate": 6
}


def convert_education(value):

    value = str(value).strip()

    if value in education_mapping:
        return education_mapping[value]

    raise ValueError(
        f"Unknown education value: {value}"
    )


# =========================================================
# 10. YES / NO CONVERSION
# =========================================================

def yes_no(value):

    if pd.isna(value):

        raise ValueError(
            "A Yes/No field contains a missing value."
        )

    value = str(value).strip().lower()

    if value == "yes":
        return 1

    if value == "no":
        return 0

    # Also allow numeric values
    if value in ["1", "1.0"]:
        return 1

    if value in ["0", "0.0"]:
        return 0

    raise ValueError(
        f"Unknown Yes/No value: {value}"
    )


# =========================================================
# 11. GENERAL HEALTH CONVERSION
# =========================================================

general_health_mapping = {

    "Excellent": 1,
    "Very good": 2,
    "Good": 3,
    "Fair": 4,
    "Poor": 5
}


def convert_general_health(value):

    if pd.isna(value):

        raise ValueError(
            "General Health contains a missing value."
        )

    # Numeric value
    if isinstance(
        value,
        (
            int,
            float,
            np.integer,
            np.floating
        )
    ):

        numeric_value = int(value)

        if 1 <= numeric_value <= 5:
            return numeric_value


    # Numeric string
    value_str = str(value).strip()

    if value_str.isdigit():

        numeric_value = int(value_str)

        if 1 <= numeric_value <= 5:
            return numeric_value


    # Text value
    if value_str in general_health_mapping:

        return general_health_mapping[value_str]


    raise ValueError(
        f"Unknown general health value: {value}"
    )


# =========================================================
# 12. INCOME CONVERSION
# =========================================================
#
# User-entered income:
#       MONTHLY INDIAN RUPEES
#
# Model's income:
#       ANNUAL US DOLLAR BRFSS CATEGORY
#
# Conversion:
#
# Monthly INR
#      ↓
# Annual INR
#      ↓
# Annual USD
#      ↓
# BRFSS Income Category 1–8
#
# =========================================================

INR_PER_USD = 96.76


def extract_monthly_inr(value):

    if pd.isna(value):

        raise ValueError(
            "Income contains a missing value."
        )

    value = str(value).strip()

    # Remove currency symbol and commas
    clean_value = (
        value
        .replace("₹", "")
        .replace(",", "")
        .strip()
    )

    # Extract numbers
    numbers = re.findall(
        r"\d+(?:\.\d+)?",
        clean_value
    )

    numbers = [
        float(number)
        for number in numbers
    ]


    if len(numbers) == 0:

        raise ValueError(
            f"Could not extract income from: {value}"
        )


    # Below ₹10,000
    if "Below" in value:

        return numbers[0]


    # ₹75,000 or more
    if "or more" in value:

        return numbers[0]


    # Range such as:
    # ₹25,000–₹35,000
    #
    # Use midpoint as the representative
    # monthly income for conversion.

    if len(numbers) >= 2:

        return (
            numbers[0] +
            numbers[1]
        ) / 2


    return numbers[0]


def convert_income(value):

    # -----------------------------------------------------
    # 1. Monthly INR
    # -----------------------------------------------------

    monthly_inr = extract_monthly_inr(
        value
    )


    # -----------------------------------------------------
    # 2. Monthly INR → Annual INR
    # -----------------------------------------------------

    annual_inr = (
        monthly_inr * 12
    )


    # -----------------------------------------------------
    # 3. Annual INR → Annual USD
    # -----------------------------------------------------

    annual_usd = (
        annual_inr /
        INR_PER_USD
    )


    # -----------------------------------------------------
    # 4. Annual USD → BRFSS Income Code
    # -----------------------------------------------------

    if annual_usd < 10000:

        income_code = 1

    elif annual_usd < 15000:

        income_code = 2

    elif annual_usd < 20000:

        income_code = 3

    elif annual_usd < 25000:

        income_code = 4

    elif annual_usd < 35000:

        income_code = 5

    elif annual_usd < 50000:

        income_code = 6

    elif annual_usd < 75000:

        income_code = 7

    else:

        income_code = 8


    return income_code


# =========================================================
# 13. INCOME CONVERSION AUDIT
# =========================================================

income_audit = pd.DataFrame()

income_audit["Original Income"] = df[
    income_col
].values

income_audit["Monthly INR"] = df[
    income_col
].apply(
    extract_monthly_inr
).values

income_audit["Annual INR"] = (
    income_audit["Monthly INR"] * 12
)

income_audit["Annual USD"] = (
    income_audit["Annual INR"] /
    INR_PER_USD
)

income_audit["BRFSS Income Code"] = df[
    income_col
].apply(
    convert_income
).values


print("\n" + "=" * 90)
print("INCOME CONVERSION AUDIT")
print("=" * 90)

display(
    income_audit.style.format({
        "Monthly INR": "₹{:,.2f}",
        "Annual INR": "₹{:,.2f}",
        "Annual USD": "${:,.2f}"
    })
)


# =========================================================
# 14. CREATE PROCESSED DATAFRAME
# =========================================================

processed = pd.DataFrame(
    index=df.index
)


# =========================================================
# 15. AGE
# =========================================================

processed["Age"] = df[
    age_col
].apply(
    convert_age
)


# =========================================================
# 16. SEX
# =========================================================

processed["Sex"] = df[
    sex_col
].apply(
    convert_sex
)


# =========================================================
# 17. EDUCATION
# =========================================================

processed["Education"] = df[
    education_col
].apply(
    convert_education
)


# =========================================================
# 18. INCOME
# =========================================================

processed["Income"] = df[
    income_col
].apply(
    convert_income
)


# =========================================================
# 19. HEIGHT + WEIGHT → BMI
# =========================================================

height_cm = pd.to_numeric(
    df[height_col],
    errors="coerce"
)

weight_kg = pd.to_numeric(
    df[weight_col],
    errors="coerce"
)


height_m = (
    height_cm / 100
)


processed["BMI"] = (
    weight_kg /
    (height_m ** 2)
)


# =========================================================
# 20. HEALTH FEATURES
# =========================================================

processed["HighBP"] = df[
    highbp_col
].apply(
    yes_no
)

processed["HighChol"] = df[
    highchol_col
].apply(
    yes_no
)

processed["CholCheck"] = df[
    cholcheck_col
].apply(
    yes_no
)

processed["Smoker"] = df[
    smoker_col
].apply(
    yes_no
)

processed["Stroke"] = df[
    stroke_col
].apply(
    yes_no
)

processed["HeartDiseaseorAttack"] = df[
    heart_col
].apply(
    yes_no
)

processed["PhysActivity"] = df[
    physical_activity_col
].apply(
    yes_no
)

processed["Fruits"] = df[
    fruit_col
].apply(
    yes_no
)

processed["Veggies"] = df[
    vegetable_col
].apply(
    yes_no
)

processed["HvyAlcoholConsump"] = df[
    alcohol_col
].apply(
    yes_no
)

processed["AnyHealthcare"] = df[
    healthcare_col
].apply(
    yes_no
)

processed["NoDocbcCost"] = df[
    doctor_cost_col
].apply(
    yes_no
)


# =========================================================
# 21. GENERAL HEALTH
# =========================================================

processed["GenHlth"] = df[
    general_health_col
].apply(
    convert_general_health
)


# =========================================================
# 22. MENTAL HEALTH
# =========================================================

processed["MentHlth"] = pd.to_numeric(
    df[mental_health_col],
    errors="coerce"
)


# =========================================================
# 23. PHYSICAL HEALTH
# =========================================================

processed["PhysHlth"] = pd.to_numeric(
    df[physical_health_col],
    errors="coerce"
)


# =========================================================
# 24. DIFFICULTY WALKING
# =========================================================

processed["DiffWalk"] = df[
    diffwalk_col
].apply(
    yes_no
)


# =========================================================
# 25. EXACT 21 MODEL FEATURES
# =========================================================

feature_names = [

    "HighBP",
    "HighChol",
    "CholCheck",
    "BMI",
    "Smoker",
    "Stroke",
    "HeartDiseaseorAttack",
    "PhysActivity",
    "Fruits",
    "Veggies",
    "HvyAlcoholConsump",
    "AnyHealthcare",
    "NoDocbcCost",
    "GenHlth",
    "MentHlth",
    "PhysHlth",
    "DiffWalk",
    "Sex",
    "Age",
    "Education",
    "Income"
]


X_unseen = processed[
    feature_names
].copy()


# =========================================================
# 26. ACTUAL DIABETES LABEL
# =========================================================

y_actual = df[
    actual_diabetes_col
].apply(
    yes_no
)


# =========================================================
# 27. VALIDATE DATA
# =========================================================

print("\n" + "=" * 90)
print("PREPROCESSING VALIDATION")
print("=" * 90)

print(
    "Respondents:",
    len(X_unseen)
)

print(
    "Model features:",
    X_unseen.shape[1]
)


print("\nFeature order:")

for number, feature in enumerate(
    X_unseen.columns,
    start=1
):

    print(
        f"{number:02d}. {feature}"
    )


# =========================================================
# 28. MISSING VALUE CHECK
# =========================================================

missing_values = X_unseen.isnull().sum()


if missing_values.sum() > 0:

    print("\nMissing values found:")

    display(
        missing_values[
            missing_values > 0
        ]
    )

    raise ValueError(
        "Missing values must be handled before prediction."
    )

else:

    print(
        "\nNo missing values found."
    )


# =========================================================
# 29. DISPLAY PROCESSED DATA
# =========================================================

print("\n" + "=" * 90)
print("PROCESSED MODEL INPUT")
print("=" * 90)

display(
    X_unseen
)


# =========================================================
# 30. MODEL PREDICTIONS + PROBABILITIES
# =========================================================

results = pd.DataFrame()

results["Actual"] = y_actual.values


for name, model_data in models.items():

    print(f"\nRunning {name}...")

    # -----------------------------------------------------
    # Extract the actual trained model
    # -----------------------------------------------------

    model = model_data["model"]

    # -----------------------------------------------------
    # Prepare model input
    # -----------------------------------------------------

    if name == "Logistic Regression":

        model_input = scaler.transform(
            X_unseen
        )

    else:

        model_input = X_unseen

    # -----------------------------------------------------
    # Prediction
    # -----------------------------------------------------

    predictions = model.predict(
        model_input
    )

    results[name] = predictions

    # -----------------------------------------------------
    # Probability
    # -----------------------------------------------------

    if hasattr(
        model,
        "predict_proba"
    ):

        probabilities = model.predict_proba(
            model_input
        )


        # Determine which probability column
        # corresponds to class 1 = Diabetes.

        if hasattr(
            model,
            "classes_"
        ):

            classes = list(
                model.classes_
            )

            if 1 in classes:

                diabetes_index = classes.index(1)

            else:

                raise ValueError(
                    f"{name} does not contain class 1."
                )

        else:

            diabetes_index = 1


        results[
            f"{name}_NoDiabetes_Probability"
        ] = probabilities[:, 0]


        results[
            f"{name}_Diabetes_Probability"
        ] = probabilities[
            :,
            diabetes_index
        ]


print(
    "\nAll models completed prediction."
)


# =========================================================
# 31. READABLE PREDICTION LABELS
# =========================================================

results["Actual Label"] = results[
    "Actual"
].map({

    0: "No Diabetes",

    1: "Diabetes"
})


for name in models.keys():

    results[
        f"{name}_Label"
    ] = results[
        name
    ].map({

        0: "No Diabetes",

        1: "Diabetes"
    })


# =========================================================
# 32. CLEAN PROBABILITY TABLE
# =========================================================
#
# This is the important part for proper alignment.
#
# Short column names are used instead of long names.
# Jupyter will render this as a proper table.
# =========================================================

probability_table = pd.DataFrame()


probability_table["Respondent"] = np.arange(
    1,
    len(results) + 1
)


probability_table["Actual"] = results[
    "Actual Label"
]


# ---------------------------------------------------------
# Logistic Regression
# ---------------------------------------------------------

probability_table["LR Prediction"] = results[
    "Logistic Regression_Label"
]

probability_table["LR Diabetes %"] = (
    results[
        "Logistic Regression_Diabetes_Probability"
    ] * 100
)


# ---------------------------------------------------------
# Decision Tree
# ---------------------------------------------------------

probability_table["DT Prediction"] = results[
    "Decision Tree_Label"
]

probability_table["DT Diabetes %"] = (
    results[
        "Decision Tree_Diabetes_Probability"
    ] * 100
)


# ---------------------------------------------------------
# Random Forest
# ---------------------------------------------------------

probability_table["RF Prediction"] = results[
    "Random Forest_Label"
]

probability_table["RF Diabetes %"] = (
    results[
        "Random Forest_Diabetes_Probability"
    ] * 100
)


# ---------------------------------------------------------
# XGBoost
# ---------------------------------------------------------

probability_table["XGB Prediction"] = results[
    "XGBoost_Label"
]

probability_table["XGB Diabetes %"] = (
    results[
        "XGBoost_Diabetes_Probability"
    ] * 100
)


# =========================================================
# 33. ROUND PROBABILITIES
# =========================================================

probability_columns = [

    "LR Diabetes %",
    "DT Diabetes %",
    "RF Diabetes %",
    "XGB Diabetes %"

]


probability_table[
    probability_columns
] = probability_table[
    probability_columns
].round(2)


# =========================================================
# 34. DISPLAY CLEAN PROBABILITY TABLE
# =========================================================

print("\n" + "=" * 120)
print("MODEL PREDICTIONS AND DIABETES PROBABILITIES")
print("=" * 120)

display(
    probability_table.style
    .format({

        "LR Diabetes %":
            "{:.2f}%",

        "DT Diabetes %":
            "{:.2f}%",

        "RF Diabetes %":
            "{:.2f}%",

        "XGB Diabetes %":
            "{:.2f}%"
    })
    .set_properties(
        **{
            "text-align": "center"
        }
    )
)


# =========================================================
# 35. MODEL PERFORMANCE
# =========================================================

evaluation_results = []


for name in models.keys():

    y_pred = results[name]


    accuracy = accuracy_score(
        y_actual,
        y_pred
    )


    precision = precision_score(
        y_actual,
        y_pred,
        zero_division=0
    )


    recall = recall_score(
        y_actual,
        y_pred,
        zero_division=0
    )


    f1 = f1_score(
        y_actual,
        y_pred,
        zero_division=0
    )


    evaluation_results.append({

        "Model": name,

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1-Score": f1

    })


evaluation_df = pd.DataFrame(
    evaluation_results
)


# =========================================================
# 36. MODEL COMPARISON
# =========================================================

print("\n" + "=" * 90)
print("MODEL PERFORMANCE ON UNSEEN DATA")
print("=" * 90)

display(
    evaluation_df.style
    .format({

        "Accuracy": "{:.2%}",

        "Precision": "{:.2%}",

        "Recall": "{:.2%}",

        "F1-Score": "{:.2%}"

    })
    .set_properties(
        **{
            "text-align": "center"
        }
    )
)


# =========================================================
# 37. CONFUSION MATRICES
# =========================================================

for name in models.keys():

    cm = confusion_matrix(
        y_actual,
        results[name]
    )


    cm_df = pd.DataFrame(

        cm,

        index=[
            "Actual No Diabetes",
            "Actual Diabetes"
        ],

        columns=[
            "Predicted No Diabetes",
            "Predicted Diabetes"
        ]

    )


    print("\n" + "=" * 90)

    print(
        f"CONFUSION MATRIX - {name}"
    )

    print("=" * 90)


    display(
        cm_df
    )


# =========================================================
# 38. CLASSIFICATION REPORTS
# =========================================================

for name in models.keys():

    print("\n" + "=" * 90)

    print(
        f"CLASSIFICATION REPORT - {name}"
    )

    print("=" * 90)


    report = classification_report(

        y_actual,

        results[name],

        target_names=[
            "No Diabetes",
            "Diabetes"
        ],

        zero_division=0

    )


    print(report)


# =========================================================
# 39. CREATE COMPLETE OUTPUT
# =========================================================

final_output = pd.concat(

    [

        df.reset_index(
            drop=True
        ),

        income_audit.reset_index(
            drop=True
        ),

        X_unseen.reset_index(
            drop=True
        ),

        results.reset_index(
            drop=True
        )

    ],

    axis=1
)
```

```python
import os
import joblib
import pandas as pd
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# ============================================================
# 1. FILE PATHS
# ============================================================

data_path = "Unseen Test Data.csv"

logistic_path = "logistic_regression.pkl"
decision_tree_path = "decision_tree.pkl"
random_forest_path = "random_forest.pkl"
xgboost_path = "xgboost.pkl"
scaler_path = "scaler.pkl"


# ============================================================
# 2. CHECK FILES
# ============================================================

files = {
    "External Dataset": data_path,
    "Logistic Regression": logistic_path,
    "Decision Tree": decision_tree_path,
    "Random Forest": random_forest_path,
    "XGBoost": xgboost_path,
    "Scaler": scaler_path
}

print("=" * 70)
print("FILE CHECK")
print("=" * 70)

for name, path in files.items():
    status = "FOUND" if os.path.exists(path) else "NOT FOUND"
    print(f"{name:22s}: {status}")

if not all(os.path.exists(path) for path in files.values()):
    raise FileNotFoundError(
        "\nOne or more required files were not found."
    )


# ============================================================
# 3. LOAD EXTERNAL DATASET
# ============================================================

df_test = pd.read_csv(data_path)

print("\n" + "=" * 70)
print("EXTERNAL VALIDATION DATASET")
print("=" * 70)

print("Dataset shape:", df_test.shape)


# ============================================================
# 4. MODEL FEATURES
# ============================================================

model_features = [
    "HighBP",
    "HighChol",
    "CholCheck",
    "BMI",
    "Smoker",
    "Stroke",
    "HeartDiseaseorAttack",
    "PhysActivity",
    "Fruits",
    "Veggies",
    "HvyAlcoholConsump",
    "AnyHealthcare",
    "NoDocbcCost",
    "GenHlth",
    "MentHlth",
    "PhysHlth",
    "DiffWalk",
    "Sex",
    "Age",
    "Education",
    "Income"
]

target_column = "Diabetes_012"


# ============================================================
# 5. CHECK COLUMNS
# ============================================================

missing_features = [
    feature
    for feature in model_features
    if feature not in df_test.columns
]

if missing_features:
    raise ValueError(
        f"Missing features: {missing_features}"
    )

if target_column not in df_test.columns:
    raise ValueError(
        f"Missing target column: {target_column}"
    )

print("All 21 model features are present.")
print("Target column is present.")


# ============================================================
# 6. CHECK 50/50 BALANCE
# ============================================================

print("\n" + "=" * 70)
print("TARGET DISTRIBUTION")
print("=" * 70)

print(
    df_test[target_column]
    .value_counts()
    .sort_index()
)

non_diabetic_count = (
    df_test[target_column] == 0
).sum()

diabetic_count = (
    df_test[target_column] == 1
).sum()

print("\nNon-Diabetic :", non_diabetic_count)
print("Diabetic     :", diabetic_count)


# ============================================================
# 7. CREATE X AND y
# ============================================================

X_test = df_test[
    model_features
].copy()

y_test = df_test[
    target_column
].copy()

print("\nX_test:", X_test.shape)
print("y_test:", y_test.shape)


# ============================================================
# 8. LOAD THE MODEL DICTIONARIES
# ============================================================

lr_data = joblib.load(logistic_path)
dt_data = joblib.load(decision_tree_path)
rf_data = joblib.load(random_forest_path)
xgb_data = joblib.load(xgboost_path)

scaler = joblib.load(scaler_path)


# ============================================================
# 9. EXTRACT THE ACTUAL MODELS
# ============================================================

lr_model = lr_data["model"]
dt_model = dt_data["model"]
rf_model = rf_data["model"]
xgb_model = xgb_data["model"]


# ============================================================
# 10. GET SAVED FEATURE NAMES
# ============================================================

feature_names = rf_data["feature_names"]

print("\n" + "=" * 70)
print("FEATURES USED BY THE TRAINED MODELS")
print("=" * 70)

print(feature_names)

if feature_names != model_features:
    print("\nWARNING: Feature order does not match.")
else:
    print("\n✓ Feature names and order match.")


# ============================================================
# 11. APPLY THE SAVED SCALER
# ============================================================
# IMPORTANT:
# Do NOT fit the scaler again on the external data.
# The scaler uses the parameters learned from the training data.

X_test_scaled = scaler.transform(X_test)

print("\nScaled X_test shape:", X_test_scaled.shape)


# ============================================================
# 12. STORE MODELS
# ============================================================

models = {
    "Logistic Regression": lr_model,
    "Decision Tree": dt_model,
    "Random Forest": rf_model,
    "XGBoost": xgb_model
}


# ============================================================
# 13. TEST ALL FOUR MODELS
# ============================================================

results = []

predictions = {}

probabilities = {}

confusion_matrices = {}


for model_name, model in models.items():

    print("\n" + "=" * 70)
    print("Testing:", model_name)
    print("=" * 70)

    # --------------------------------------------------------
    # Logistic Regression uses the saved scaler.
    # Tree-based models use original feature values.
    # --------------------------------------------------------

    if model_name == "Logistic Regression":

        X_input = X_test_scaled

    else:

        X_input = X_test

    # --------------------------------------------------------
    # Prediction
    # --------------------------------------------------------

    y_pred = model.predict(X_input)

    predictions[model_name] = y_pred

    # --------------------------------------------------------
    # Probability
    # --------------------------------------------------------

    if hasattr(model, "predict_proba"):

        y_prob = model.predict_proba(
            X_input
        )[:, 1]

        probabilities[model_name] = y_prob

    else:

        y_prob = None

    # --------------------------------------------------------
    # Overall metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )

    if y_prob is not None:

        roc_auc = roc_auc_score(
            y_test,
            y_prob
        )

    else:

        roc_auc = np.nan

    # --------------------------------------------------------
    # Confusion matrix
    # --------------------------------------------------------

    cm = confusion_matrix(
        y_test,
        y_pred
    )

    confusion_matrices[model_name] = cm

    tn, fp, fn, tp = cm.ravel()

    # --------------------------------------------------------
    # Non-Diabetic class = 0
    # --------------------------------------------------------

    non_diabetic_precision = precision_score(
        y_test,
        y_pred,
        pos_label=0,
        zero_division=0
    )

    non_diabetic_recall = recall_score(
        y_test,
        y_pred,
        pos_label=0,
        zero_division=0
    )

    # --------------------------------------------------------
    # Diabetic class = 1
    # --------------------------------------------------------

    diabetic_precision = precision_score(
        y_test,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    diabetic_recall = recall_score(
        y_test,
        y_pred,
        pos_label=1,
        zero_division=0
    )

    # --------------------------------------------------------
    # Store results
    # --------------------------------------------------------

    results.append({

        "Model": model_name,

        "Accuracy (%)":
            accuracy * 100,

        "Precision (%)":
            precision * 100,

        "Recall (%)":
            recall * 100,

        "F1-Score (%)":
            f1 * 100,

        "ROC-AUC (%)":
            roc_auc * 100,

        "Non-Diabetic Precision (%)":
            non_diabetic_precision * 100,

        "Non-Diabetic Recall (%)":
            non_diabetic_recall * 100,

        "Diabetic Precision (%)":
            diabetic_precision * 100,

        "Diabetic Recall (%)":
            diabetic_recall * 100,

        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })


# ============================================================
# 14. CREATE RESULTS DATAFRAME
# ============================================================

results_df = pd.DataFrame(
    results
).round(2)


# ============================================================
# 15. OVERALL COMPARISON
# ============================================================

print("\n\n" + "=" * 70)
print("OVERALL EXTERNAL VALIDATION RESULTS")
print("=" * 70)

comparison_table = results_df[
    [
        "Model",
        "Accuracy (%)",
        "Precision (%)",
        "Recall (%)",
        "F1-Score (%)",
        "ROC-AUC (%)"
    ]
]

display(comparison_table)


# ============================================================
# 16. CLASS-WISE PERFORMANCE
# ============================================================

print("\n" + "=" * 70)
print("CLASS-WISE PERFORMANCE")
print("=" * 70)

classwise_table = results_df[
    [
        "Model",
        "Non-Diabetic Precision (%)",
        "Non-Diabetic Recall (%)",
        "Diabetic Precision (%)",
        "Diabetic Recall (%)"
    ]
]

display(classwise_table)


# ============================================================
# 17. CONFUSION MATRIX VALUES
# ============================================================

print("\n" + "=" * 70)
print("CONFUSION MATRIX VALUES")
print("=" * 70)

confusion_table = results_df[
    [
        "Model",
        "TN",
        "FP",
        "FN",
        "TP"
    ]
]

display(confusion_table)


# ============================================================
# 18. CLASSIFICATION REPORTS
# ============================================================

for model_name in models:

    print("\n" + "=" * 70)
    print(model_name)
    print("=" * 70)

    print(
        classification_report(
            y_test,
            predictions[model_name],
            target_names=[
                "Non-Diabetic",
                "Diabetic"
            ],
            digits=4,
            zero_division=0
        )
    )


# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print("External dataset      :", df_test.shape)
print("Non-diabetic samples  :", non_diabetic_count)
print("Diabetic samples      :", diabetic_count)
print("Number of features    :", len(model_features))
print("Models tested         :", len(models))
print("\n✓ External validation completed successfully.")
```
